### Objetivo

Con la informacion anterior, se tiene que el proyecto se centra en el desarrollo de un modelo de riesgo crediticio basado en datos detallados sobre transacciones financieras y factores de riesgo asociados a las billeteras de los clientes. El conjunto de datos incluye información sobre las transacciones, tales como los montos prestados y reembolsados, la actividad de las billeteras, y diversos indicadores de riesgo del mercado. Además, se consideran variables como el factor de riesgo, la cantidad total prestada en Ethereum, la liquidación de activos, y los indicadores de rendimiento del mercado.

En este proyecto, se intenta predecir la probabilidad de morosidad de los clientes del Banco de Comisiones Infinitas. Específicamente, el objetivo del modelo es identificar qué clientes tienen mayor riesgo de incumplir sus obligaciones crediticias (entrar en mora), permitiendo al banco tomar decisiones informadas sobre la gestión de los préstamos. La variable target en este caso es la que indica el estado de morosidad del cliente. Esta variable objetivo tiene como valor binario si el cliente está en mora o no.

### Modelos y metricas a usar

En este trabajo, el enfoque para la construcción del modelo predictivo incluye una variedad de enfoques. Para establecer una línea base, comenzaremos con un modelo de Decision Tree, conocido por su simplicidad y capacidad de interpretar resultados rápidamente. Sin embargo, para obtener un rendimiento más robusto, se han seleccionado modelos más sofisticados como XGBoost, Extra Trees, y Random Forest, los cuales son particularmente efectivos en problemas de clasificación, dada su capacidad para manejar datos complejos y evitar el sobreajuste a través de técnicas como el boosting y el bagging. Además, utilizaremos PyCaret, una librería que nos permitirá obtener un benchmark más extenso de diferentes algoritmos, lo que facilitará la evaluación comparativa de múltiples enfoques con menor esfuerzo manual.

En cuanto a las métricas de evaluación, hemos seleccionado el F1 Score y el AUC (Área Bajo la Curva ROC). El F1 Score es adecuado en contextos donde el balance entre precisión y exhaustividad es crucial, especialmente cuando los datos están desbalanceados, como es el caso en nuestro problema. El AUC, por su parte, es ampliamente utilizado para medir el rendimiento general de los modelos en términos de su capacidad para diferenciar entre clases. El AUC es relevante no solo por su fundamento estadístico en la evaluación del rendimiento de clasificadores, sino también porque es la métrica utilizada en la competencia interna para comparar los modelos. Esto asegura que nuestros resultados sean consistentes con las expectativas del entorno competitivo en el que estamos trabajando.

El preprocesamiento de los datos sigue un enfoque estructurado para optimizar el uso de las distintas variables. Primero, se identificaron y convirtieron las variables categóricas mediante One-Hot Encoding, ya que trabajar con representaciones categóricas claras mejora el rendimiento del modelo. Luego, se excluyeron columnas destinadas solo a la identificación del dataset, ya que no aportaban valor predictivo. Las variables temporales fueron transformadas a formato datetime, permitiendo la extracción de información relevante, como mes o día de la semana, para capturar patrones temporales útiles. Finalmente, las variables numéricas fueron normalizadas para evitar que magnitudes desproporcionadas influyan en los modelos de aprendizaje automático. Este pipeline garantiza que los datos estén preparados para obtener el mejor rendimiento posible en los modelos de clasificación utilizados.

```python
def exploratory_data_analysis(dataframe):
    pd.options.display.max_rows = 200
    pd.set_option('display.max_columns', None)

    print('1.- El DataFrame tiene', dataframe.shape[0],'filas y', dataframe.shape[1], 'columnas\n')
    print('2.- El DataFrame está compuesto por las siguientes columnas: ', list(dataframe.columns),'\n')
    print('3.- Ejemplos de filas del DataFrame:\n')
    print('Primeras 5 filas:')
    display(dataframe.head(5))
    print('Últimas 5 filas:')
    display(dataframe.tail(5))
    print('Muestreo aleatorio de 5 filas:')
    display(dataframe.sample(5))
    print('4.- Descripción numérica del Dataframe:\n')
    display(dataframe.describe())
    print('5.- Cantidad de valores nulos por columna:')
    display(dataframe.isna().sum())
    print('6.- Cantidad de valores únicos por columna:')
    unique_values = dataframe.nunique()
    display(unique_values)
    print('7.- Tipos de variables por columna:')
    display(dataframe.dtypes)
    print('Desgloce variables con a lo más 10 categorías:')
    for column in dataframe.columns:
        if unique_values[column] <= 10:
            print(f'Columna: {column}')
            print(dataframe[column].value_counts())
            print('\n')
```

```python
# Configuración de logging
logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')

#definimos función eda

def eda(dataframe):
    """
    Realiza un análisis exploratorio de datos (EDA) en un DataFrame de pandas.

    Parámetros:
    - dataframe (pd.DataFrame): El DataFrame a analizar.

    Retorna:
    - None (imprime los resultados del análisis en consola).
    """
    try:
        # Configuración de pandas para la visualización
        pd.options.display.max_rows = 200
        pd.set_option('display.max_columns', None)

        # Dimensiones del DataFrame
        logging.info(f'1.- El DataFrame tiene {dataframe.shape[0]} filas y {dataframe.shape[1]} columnas\n')

        # Columnas del DataFrame
        logging.info(f'2.- Columnas del DataFrame: {list(dataframe.columns)}\n')

        # Ejemplos de filas
        logging.info('3.- Ejemplos de filas del DataFrame:\n')
        logging.info('Primeras 5 filas:')
        display(dataframe.head(5))
        logging.info('Últimas 5 filas:')
        display(dataframe.tail(5))
        logging.info('Muestreo aleatorio de 5 filas:')
        display(dataframe.sample(5, random_state=SEED))

        # Descripción numérica
        logging.info('4.- Descripción numérica del DataFrame:')
        display(dataframe.describe(include='all'))

        # Valores nulos
        logging.info('5.- Cantidad de valores nulos por columna:')
        display(dataframe.isna().sum())

        # Valores únicos
        logging.info('6.- Cantidad de valores únicos por columna:')
        unique_values = dataframe.nunique()
        display(unique_values)

        # Tipos de datos
        logging.info('7.- Tipos de datos por columna:')
        display(dataframe.dtypes)

        # Filas duplicadas
        logging.info('8.- Cantidad de filas duplicadas en el DataFrame:')
        logging.info(f'{dataframe.duplicated().sum()} filas duplicadas\n')

        # Variables categóricas con <= 6 categorías
        logging.info('9.- Distribución de variables categóricas con a lo más 6 categorías:')
        for column in dataframe.columns:
            if unique_values[column] <= 6:
                logging.info(f'\nColumna: {column}')
                display(dataframe[column].value_counts())
                display(dataframe[column].value_counts(normalize=True) * 100)

    except Exception as e:
        logging.error(f"Error durante el EDA: {e}")
```

```python
def explorar_dataframe_estado(df):
    """
    Explora el DataFrame y genera violin plots para las variables numéricas en función de la variable objetivo (target).

    Parameters:
        df (pd.DataFrame): Dataframe asociado para generar los plots
    """

    num_cols = df.select_dtypes(include=['float64', 'int64']).columns.tolist()
    num_cols = [col for col in num_cols if col != 'target']

    if not num_cols:
        print("No se encontraron columnas numéricas en el DataFrame.")
        return

    # Crear gráficos de violin plots
    nrows = len(num_cols)
    ncols = 1 if nrows <= 5 else 2
    plt.figure(figsize=(20, nrows * 3))

    for i, col in enumerate(num_cols):
        plt.subplot(nrows // ncols + (nrows % ncols > 0), ncols, i + 1)

        if df['target'].isnull().all() or df[col].isnull().all():
            print(f"No hay datos suficientes para graficar {col} por target.")
            continue

        df_violinplot = df[[col, 'target']].dropna()

        if df_violinplot['target'].nunique() < 2:
            print(f"No hay suficientes clases en 'target' para graficar {col}.")
            continue

        sns.violinplot(x='target', y=col, data=df_violinplot, inner="quartile")
        plt.title(f'Violin Plot de {col} por target', fontsize=12)
        plt.xlabel('target', fontsize=10)
        plt.ylabel(col, fontsize=10)

    plt.tight_layout()
    plt.show()

explorar_dataframe_estado(data_0)
```

De los violinplot, se puede observar que las variables *risk_factor* poseen una diferencia considerable en como se representan los datos en estos plots. Bajo el contexto del problema, tiene sentido de que la gente con mayor factor de riesgo sean aquellos con target 1, es decir, caeran en mora, o poseen mayor probabilidad de ello.

```python
def graficar_distribuciones_condicionales(data_0, max_variables_por_figura=5):
    """
    Grafica la distribución de todas las variables numéricas y datetime, diferenciando colores según la columna 'target'.

    Parameters:
        data_0 (pd.DataFrame): El DataFrame con las variables predictoras y la variable objetivo 'target'.
        max_variables_por_figura (int): Número máximo de variables a graficar por figura.
    """
    df = data_0.copy()

    variables_validas = df.select_dtypes(include='number').columns.tolist()
    datetime_variables = df.select_dtypes(include='datetime').columns.tolist()

    color_map = {0: 'lightblue', 1: 'red'}

    for i in range(0, len(variables_validas), max_variables_por_figura):
        plt.figure(figsize=(15, 3 * min(max_variables_por_figura, len(variables_validas) - i)))

        for j, var in enumerate(variables_validas[i:i + max_variables_por_figura]):
            if df[var].nunique() > 1:
                plt.subplot(min(max_variables_por_figura, len(variables_validas) - i), 1, j + 1)

                total_por_target = df['target'].value_counts(normalize=False)

                for target in total_por_target.index:
                    df_target = df[df['target'] == target]
                    sns.histplot(data=df_target, x=var, bins=30, kde=True, alpha=0.7,
                                 label=f'Target = {target}', stat='percent', common_bins=True,
                                 color=color_map.get(target, 'gray'))

                plt.title(f'Distribución de {var} por Target', fontsize=12)
                plt.xlabel(var, fontsize=10)
                plt.ylabel('Porcentaje (%)', fontsize=10)
                plt.grid(axis='y', alpha=0.75)
                plt.legend(title='Target')

        plt.tight_layout()
        plt.show()

    for i in range(0, len(datetime_variables), max_variables_por_figura):
        plt.figure(figsize=(15, 3 * min(max_variables_por_figura, len(datetime_variables) - i)))

        for j, var in enumerate(datetime_variables[i:i + max_variables_por_figura]):
            if df[var].nunique() > 1:
                plt.subplot(min(max_variables_por_figura, len(datetime_variables) - i), 1, j + 1)

                total_por_target = df['target'].value_counts(normalize=False)

                for target in total_por_target.index:
                    df_target = df[df['target'] == target]

                    df_target_sorted = df_target.sort_values(by=var)
                    sns.lineplot(data=df_target_sorted, x=var, y=[1]*len(df_target_sorted),
                                 label=f'Target = {target}', color=color_map.get(target, 'gray'))

                plt.title(f'Distribución temporal de {var} por Target', fontsize=12)
                plt.xlabel(var, fontsize=10)
                plt.ylabel('Frecuencia', fontsize=10)
                plt.grid(axis='y', alpha=0.75)
                plt.legend(title='Target')

        plt.tight_layout()
        plt.show()

graficar_distribuciones_condicionales(data_0)
```

#### Exploramos outliers
Como buscamos predecir casos borde veremos tambien outliers.

```python
def detect_outliers_iqr(df):
    outliers_iqr = {}
    for column in df.select_dtypes(include=np.number).columns:
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        outliers_iqr[column] = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return outliers_iqr

def detect_outliers_std(df):
    outliers_std = {}
    for column in df.select_dtypes(include=np.number).columns:
        mean = df[column].mean()
        std_dev = df[column].std()
        lower_bound = mean - 3 * std_dev
        upper_bound = mean + 3 * std_dev
        outliers_std[column] = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return outliers_std

outliers_iqr = detect_outliers_iqr(data_0)
outliers_std = detect_outliers_std(data_0)

print("Outliers detectados con IQR:")
for column, outliers in outliers_iqr.items():
    print(f"\nColumna: {column}")
    print(outliers)

print("\nOutliers detectados con Desviación Estándar:")
for column, outliers in outliers_std.items():
    print(f"\nColumna: {column}")
    print(outliers)
```

## Preprocesamiento

**Explicación:**
1. **Conversión a `datetime`**:  
   Se usa `pd.to_datetime` para convertir las columnas de timestamps a un formato de fecha y hora de pandas. La opción `unit='s'` indica que los valores están en segundos desde la fecha base (epoch). El parámetro `errors='coerce'` asegura que si hay valores no válidos (como `999999999` o NaN), se conviertan en `NaT` (Not a Time).

2. **Extracción de características temporales**:  
   Para cada columna temporal convertida, se extraen las siguientes características:
   - **Año (`_year`)**
   - **Mes (`_month`)**
   - **Día (`_day`)**
   - **Día de la semana (`_weekday`)**: Representado como un número de 0 (lunes) a 6 (domingo).

Estas nuevas columnas pueden ser útiles para el modelo, ya que pueden capturar patrones temporales importantes. Solo se realizaran con las variables sin problemas de datos, para evitar manejar mas nulos numericos.

**Consideraciones:**
- **`NaT`**: Si alguna de las columnas contiene valores `999999999` o similares, estos se convertirán a `NaT` (Not a Time).

### 2- **Pipeline con todo el preprocesamiento**

```python
from sklearn.preprocessing import OneHotEncoder, FunctionTransformer

# Paso 1: Definir las transformaciones previas
categorical = ['borrow_block_number', 'wallet_address', 'market_ht_trendmode']
datetime_columns = ['borrow_timestamp', 'first_tx_timestamp', 'last_tx_timestamp',
                    'risky_first_tx_timestamp', 'risky_last_tx_timestamp']

datetime_columns_features = ['borrow_timestamp', 'first_tx_timestamp', 'last_tx_timestamp']

columnas_a_excluir = [
    'borrow_block_number', 'wallet_address', 'time_since_last_liquidated',
    'market_ht_trendmode', 'borrow_timestamp', 'first_tx_timestamp',
    'last_tx_timestamp', 'risky_first_tx_timestamp', 'risky_last_tx_timestamp',
    'risky_first_last_tx_timestamp_diff', 'time_since_first_deposit',
    'market_ppo', 'market_rocp', "market_rocr", 'target'
]

# Funciones de preprocesamiento
def convert_to_category(df):
    df[categorical] = df[categorical].astype('category')
    return df

def process_datetime_columns(df):
    df['risky_first_tx_timestamp'] = df['risky_first_tx_timestamp'].replace(999999999, pd.NA)
    df['risky_last_tx_timestamp'] = df['risky_last_tx_timestamp'].replace(999999999, pd.NA)
    for col in datetime_columns:
        df[col] = pd.to_datetime(df[col], unit='s', errors='coerce')
    return df

def extract_temporal_features(df):
    for col in datetime_columns_features:
        if col in df:
            df[f'{col}_year'] = df[col].dt.year
            df[f'{col}_month'] = df[col].dt.month
            df[f'{col}_day'] = df[col].dt.day
            df[f'{col}_weekday'] = df[col].dt.weekday
    return df

def normalize_columns(df):
    cols_a_normalizar = [col for col in df.select_dtypes(include=[np.number]).columns if col not in columnas_a_excluir]
    df_normalized = df.copy()
    for col in cols_a_normalizar:
        mean_val = df_normalized[col].mean()
        std_val = df_normalized[col].std()
        df_normalized[col] = (df_normalized[col] - mean_val) / std_val
    return df_normalized

def exclude_columns(df):
    df_excluded = df.drop(['borrow_block_number', 'wallet_address', "time_since_last_liquidated"], axis=1)
    return df_excluded

def one_hot_encode(df):
    ohe = OneHotEncoder(sparse_output=False)
    encoded = ohe.fit_transform(df[['market_ht_trendmode']])
    encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(['market_ht_trendmode']))
    df = df.drop('market_ht_trendmode', axis=1).reset_index(drop=True)
    return pd.concat([df, encoded_df], axis=1)

def remove_datetime_columns(df):
    return df.select_dtypes(exclude=['datetime64[ns]'])

# Paso 2: Crear transformadores personalizados
categorical_transformer = FunctionTransformer(convert_to_category, validate=False)
datetime_transformer = FunctionTransformer(process_datetime_columns, validate=False)
temporal_feature_extractor = FunctionTransformer(extract_temporal_features, validate=False)
normalization_transformer = FunctionTransformer(normalize_columns, validate=False)
exclude_transformer = FunctionTransformer(exclude_columns, validate=False)
one_hot_transformer = FunctionTransformer(one_hot_encode, validate=False)
remove_datetime_transformer = FunctionTransformer(remove_datetime_columns, validate=False)

# Pipeline en el orden solicitado con extracción de características temporales
preprocessing_pipeline = Pipeline([
    ('categorical', categorical_transformer),              # Convertimos a categóricas si es necesario
    ('exclude_columns', exclude_transformer),              # Excluimos columnas específicas
    ('one_hot_encoding', one_hot_transformer),             # Aplicamos One-Hot Encoding
    ('datetime', datetime_transformer),                    # Convertimos las columnas datetime
    ('temporal_features', temporal_feature_extractor),     # Extraemos las características temporales
    ('datetime_exclusion', remove_datetime_transformer),   # Excluimos las columnas datetime originales
    ('normalization', normalization_transformer)           # Normalizamos las columnas numéricas
])
```

```python
class CustomPredictor(BaseEstimator, TransformerMixin):
    def __init__(self, model):
        self.model = model

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        return self.model.predict(X)

    def predict_proba(self, X):
        return self.model.predict_proba(X)

def model_report(model, data):
    data_no_target = data.drop('target', axis=1)

    unseen_predictions = predict_model(model, data=data_no_target)

    predicted_labels = unseen_predictions['prediction_label']
    true_labels = data['target']

    report = classification_report(true_labels, predicted_labels)
    predicted_probabilities = unseen_predictions['prediction_score']

    # Calcular AUC
    auc_score = roc_auc_score(true_labels, predicted_probabilities)

    # Imprimir los resultados
    print(report)
    print(f"AUC Score: {auc_score:.4f}")

    return report, auc_score
```

## Re-entrenamiento del modelo con data nueva

Como se solicita, se debe considerar el hecho de que llega data nueva, y por ello el modelo debe ser actualizado, todo esto de forma automatizada, es por ello, que se definira una funcion para ello, considerando el re-entrenamiento, junto a la optimizacion de los hiperparametros mas relevantes segun el grafico de hiperparametros mas relevantes en el desempeño del modelo, es decir, nos centraremos en optimizar *max_depth*, y *subsample*.

La logica detras de esta funcion es poder realizar re-entrenamiento del modelo con data nueva. Primeramente, se realiza el preprocesamiento mediante el pipeline creado con anterioridad en el proyecto. Luego, se optimizan hiperparametros con optuna, con el criterio antes mencionado para escoger cuales son los que se buscaria aspecto de mejora.

Tras tener la mejor combinacion de hiperparametros, se entrena el modelo con la data nueva, y luego se guarda en la misma carpeta donde estan el resto de modelos, con un diferenciador de timestamp para saber cual modelo es cual.

Pipeline de reentreno con MLFlow inlcuido

```python
def reentrenar_modelo_con_optuna_mlflow(modelo, X_new_input, y_new_input, preprocessing_pipeline, base_params, SEED=444): #suponiendo que la data nos llega en el mismo formato que nos han dado ahora, osea en archivos X y y separados
    """
    Re-entrena un modelo con Optuna y registra los experimentos usando MLFlow.
    """
    import mlflow
    import mlflow.sklearn
    import optuna
    from datetime import datetime
    from sklearn.model_selection import train_test_split
    from sklearn.metrics import roc_auc_score, f1_score, recall_score
    import joblib
    import pandas as pd
    import xgboost as xgb
    import shap
    import matplotlib.pyplot as plt
    import os

    # Verifica si el experimento ya existe
    experiment_name = "Proyecto_ML_Reentrenamiento"
    experiment = mlflow.get_experiment_by_name(experiment_name)

    if experiment is None:
        experiment = mlflow.create_experiment(experiment_name)
        print(f"Nuevo experimento creado con ID: {experiment}")
    else:
        print(f"El experimento ya existe con ID: {experiment.experiment_id}")

    mlflow.set_experiment(experiment_name)
    mlflow.set_tracking_uri("http://127.0.0.1:5000")
    print("Experimento configurado:", mlflow.get_experiment_by_name(experiment_name))

    # Procesamiento de los datos
    data_new = pd.concat([X_new_input.reset_index(drop=True), y_new_input.reset_index(drop=True)], axis=1)
    data_new_preprocessed = preprocessing_pipeline.fit_transform(data_new)

    X_new_preprocessed = data_new_preprocessed.drop('target', axis=1)
    y_new = data_new_preprocessed['target']

    X_train, X_val, y_train, y_val = train_test_split(X_new_preprocessed, y_new, test_size=0.3, random_state=SEED)

    # Generar el nombre del run con un timestamp
    timestamp = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")
    run_name = f"Re-Train_Optuna_{timestamp}"

    with mlflow.start_run(run_name=run_name):
        
        def objective(trial):
            param = base_params.copy()
            param['max_depth'] = trial.suggest_int('max_depth', 9, 15)
            param['subsample'] = trial.suggest_uniform('subsample', 0.70, 0.85)

            model = xgb.XGBClassifier(**param)
            model.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=False)

            y_pred = model.predict_proba(X_val)[:, 1]
            auc_score = roc_auc_score(y_val, y_pred)

            return auc_score

        study = optuna.create_study(direction="maximize")
        study.optimize(objective, n_trials=2)
        best_params = study.best_params
        print("Mejores parámetros encontrados:", best_params)

        final_params = {**base_params, **best_params, "objective": "binary:logistic", "eval_metric": "auc", "use_label_encoder": False}

        modelo_final = xgb.XGBClassifier(**final_params)
        modelo_final.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=False)

        # Guardar modelo
        path = "./resultados"  # Define el path si no existe
        os.makedirs(path, exist_ok=True)
        updated_model_path = f"{path}/modelo_actualizado_{timestamp}.pkl"
        joblib.dump(modelo_final, updated_model_path)
        print(f"Modelo actualizado guardado en: {updated_model_path}")

        # Predicciones
        y_pred = modelo_final.predict(X_val)
        y_proba = modelo_final.predict_proba(X_val)[:, 1]

        # Métricas adicionales
        auc_score = roc_auc_score(y_val, y_proba)
        f1 = f1_score(y_val, y_pred)
        recall = recall_score(y_val, y_pred)

        print(f"AUC: {auc_score}, F1-Score: {f1}, Recall: {recall}")

        # Registrar parámetros, métricas y modelo en MLFlow
        mlflow.log_params(final_params)
        mlflow.log_metric("AUC", auc_score)
        mlflow.log_metric("F1-Score", f1)
        mlflow.log_metric("Recall", recall)
        mlflow.sklearn.log_model(modelo_final, "Modelo Actualizado")
        mlflow.log_artifact(updated_model_path, artifact_path="modelos")

        # Gráfico de importancia de características
        plt.figure(figsize=(10, 20))
        xgb.plot_importance(modelo_final, importance_type='weight')
        plt.title("Importancia de Características")
        importance_path = f"{path}/feature_importance_{timestamp}.png"
        plt.savefig(importance_path)
        plt.close()
        mlflow.log_artifact(importance_path, artifact_path="graficos")

        # Gráfico de SHAP values
        X_val_sample = X_val.sample(n=1000, random_state=SEED)  # Se toma una submuestra para hacer más liviando el gráfico

        explainer = shap.TreeExplainer(modelo_final)
        shap_values = explainer.shap_values(X_val_sample)
       

        plt.figure(figsize=(12, 8))
        shap.summary_plot(shap_values, X_val_sample, plot_type="bar", show=False, max_display=20)

        # Guardar el grafico
        shap_path = f"{path}/shap_summary_{timestamp}.png"
        plt.savefig(shap_path, bbox_inches='tight', dpi=100)  # Ajustar la resolución para hacer más liviano
        plt.close() 
   
        mlflow.log_artifact(shap_path, artifact_path="graficos")

    return modelo_final
```
